# Fundamentos de Python - Pre-Bootcamp

## 1. ¿Por qué aprender Python?

Python es un lenguaje de programación popular por su sintaxis clara, su versatilidad y la gran cantidad de herramientas disponibles. Es una **excelente opción para comenzar** a programar.

## 2. Instalación y uso de Python

Durante esta etapa, utilizaremos **Visual Studio Code (VS Code)** como entorno de desarrollo (IDE).

### ¿Qué necesitas instalar?
- Python (https://www.python.org/downloads/)
- VS Code (https://code.visualstudio.com/)
- Extensión "Python" para VS Code

> **Ejercicio:** Instala Python y crea un nuevo archivo con extensión `.py` en VS Code. Escribe `print("Hola mundo")` en él y ejecútalo!

## 3. Sintaxis básica

````python
# Comentario en Python
nombre = "Sofía"
edad = 20
print(nombre, "tiene", edad, "años")
````

Python usa **indentación** (espacios, por lo general tabuladores) para definir bloques de código. No se usan llaves `{}` como en otros lenguajes.

## 4. Tipos de datos primitivos

````python
# Números
x = 10        # int
y = 3.14      # float

# Texto
mensaje = "Hola mundo"  # str

# Booleanos
activo = True
````

Puedes verificar el tipo de una variable con la función `type()`.

## 5. Estructuras de datos

````python
# Lista (mutable, ordenada)
frutas = ["manzana", "banana", "naranja"]

# Tupla (inmutable, ordenada)
puntos = (10, 20)

# Diccionario (clave-valor)
persona = {"nombre": "Luis", "edad": 30}
````

## 6. Operadores y expresiones

````python
# Aritméticos
a = 5 + 3
b = 10 / 2
c = 2 ** 3

# Comparación
a == b   # igual
a != b   # distinto
a < b    # menor

# Lógicos
True and False
True or False
not True
````

## 7. Condicionales

````python
edad = 18
if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
````

También se puede usar `elif` para múltiples condiciones.

## 8. Bucles

````python
# Bucle for
for i in range(5):
    print(i)

# Bucle while
contador = 0
while contador < 5:
    print(contador)
    contador += 1
````

## 9. Funciones

````python
def saludar(nombre):
    return f"Hola, {nombre}"

print(saludar("Ana"))
````

Las funciones permiten reutilizar bloques de código.

# Git y GitHub - Fundamentos

## 1. ¿Qué es Git?

**Git** es un sistema de control de versiones que te permite guardar el historial de cambios de tus archivos y volver a versiones anteriores si es necesario. Es especialmente útil cuando trabajas en proyectos de programación.

> Git se usa desde la línea de comandos. No necesitas interfaces gráficas para esta etapa inicial.

## 2. ¿Qué es GitHub?

**GitHub** es una plataforma en línea para almacenar tus repositorios Git. Permite guardar tu trabajo en la nube, colaborar con otras personas y mostrar tus proyectos.

## 3. Instalación de Git

### Windows
1. Descargar desde: https://git-scm.com/download/win
2. Ejecutar el instalador y mantener la configuración por omisión.
3. Verificar instalación:
   Abre `Git Bash` y ejecuta:
   ```bash
   git --version
   ```

### macOS
1. Abrir la terminal y ejecutar:
   ```bash
   git --version
   ```
   Si no está instalado, el sistema ofrecerá instalarlo automáticamente.

2. También puedes instalar Git con Homebrew:
   ```bash
   brew install git
   ```

### Linux (Debian/Ubuntu)
1. Ejecutar en terminal:
   ```bash
   sudo apt update
   sudo apt install git
   ```

2. Verificar instalación en terminal:
   ```bash
   git --version
   ```


## 4. Crear una cuenta en GitHub

1. Ir a [https://github.com/](https://github.com/)
2. Hacer clic en **Sign up**.
3. Seguir los pasos para crear tu usuario (email, contraseña, nombre de usuario).

> Recuerda verificar tu correo electrónico después de registrarte.

## 5. Conectar Git con tu cuenta de GitHub

### 5.1 Configura tu nombre y correo en Git

Esto lo haces una sola vez:

```bash
git config --global user.name "Tu Nombre"
git config --global user.email "tucorreo@ejemplo.com"
```

### 5.2 Generar una clave SSH (opcional, pero recomendado)

1. En tu terminal:

```bash
ssh-keygen -t ed25519 -C "tucorreo@ejemplo.com"
```

Presiona Enter varias veces para aceptar las opciones por defecto.

2. Copia la clave pública:

```bash
cat ~/.ssh/id_ed25519.pub
```

3. Ve a tu cuenta de GitHub → *Settings* → *SSH and GPG keys* → *New SSH Key*  
   Pega la clave y guarda.

> Esto permite subir código sin ingresar tu contraseña cada vez.

## 6. Crear un repositorio local

```bash
mkdir mi_proyecto
cd mi_proyecto
git init
```

Esto inicializa un repositorio Git en esa carpeta.

## 7. Flujo básico de trabajo

```bash
# Ver el estado
git status

# Agregar archivo
git add archivo.py

# Confirmar cambio
git commit -m "Mi primer commit"

# Ver historial
git log
```

## 8. Subir tu proyecto a GitHub

1. Crea un nuevo repositorio en https://github.com
2. Copia la URL (SSH o HTTPS)
3. En tu terminal:

```bash
git remote add origin git@github.com:usuario/nombre-repo.git  # si usas SSH
git branch -M main
git push -u origin main
```

## 9. Clonar un repositorio existente

```bash
git clone https://github.com/usuario/nombre-repo.git
```

Esto descargará una copia local del proyecto.

# Introducción a Metodologías Ágiles y SCRUM

## 1. ¿Qué es una metodología ágil?

Una metodología ágil es una forma de trabajar que permite desarrollar productos de forma iterativa y flexible, adaptándose a cambios frecuentes y necesidades reales.

Se basa en principios como:
- Entregar valor de forma rápida y continua.
- Colaboración entre todos los involucrados.
- Mejora continua y retroalimentación.

## 2. ¿Qué es SCRUM?

**SCRUM** es una de las metodologías ágiles más utilizadas.  
Se enfoca en dividir el trabajo en bloques cortos llamados **sprints** (de 1 a 4 semanas) para entregar avances funcionales de manera frecuente.

## 3. Roles en SCRUM

En un equipo SCRUM existen tres roles principales:

- **Product Owner (PO):** Representa al cliente. Define qué se debe hacer y en qué orden.
- **Scrum Master:** Facilita el proceso. Se asegura de que el equipo entienda y siga SCRUM.
- **Equipo de Desarrollo:** Personas encargadas de realizar el trabajo técnico.

## 4. Eventos principales

SCRUM define eventos que estructuran el trabajo:

- **Sprint:** Periodo de tiempo (por ejemplo, 2 semanas) en el que se trabaja sobre tareas específicas.
- **Sprint Planning:** Reunión para definir qué se hará en el próximo sprint.
- **Daily Stand-Up:** Reunión diaria breve (15 min) donde cada miembro responde:
  - ¿Qué hice ayer?
  - ¿Qué haré hoy?
  - ¿Hay algo que me bloquee?
- **Sprint Review:** Se presentan los avances del sprint.
- **Sprint Retrospective:** Reunión de mejora para revisar qué funcionó y qué no.

## 5. Artefactos de SCRUM

- **Product Backlog:** Lista de tareas deseadas por el cliente.
- **Sprint Backlog:** Subconjunto de tareas seleccionadas para el sprint actual.
- **Incremento:** Resultado del sprint

## 6. Herramientas ágiles recomendadas

Puedes organizar tareas y flujos con herramientas como:

- https://trello.com/
- https://github.com/features/project-management
- https://www.atlassian.com/software/jira

## 7. Actividad sugerida

Trabaja en grupo de 3 personas. Cada integrante debe colaborar en un tablero compartido usando **Trello** (https://trello.com/).

1. Un integrante del grupo crea un tablero con tres columnas: `Por hacer`, `En progreso` y `Hecho`.
2. Invita a los demás miembros del equipo al tablero.
3. Juntos, distribuyan tareas relacionadas a un pequeño proyecto (por ejemplo, automatizar algo con Python, hacer una página personal, etc.).
4. Cada persona debe mover al menos una tarjeta de una columna a otra (para simular el flujo de trabajo).

Ejemplo de estructura:

| Por hacer           | En progreso        | Hecho          |
|---------------------|--------------------|----------------|
| Crear repositorio   | Escribir README    | Instalar Git   |
| Diseñar portada     |                    |                |

> Durante la revisión, asegúrate de que cada integrante haya movido tareas y que el tablero refleje un avance visible.
