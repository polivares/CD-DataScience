# Introducción a los Servicios en la Nube

Estos servicios permiten acceder a recursos como cómputo, almacenamiento, bases de datos, etc. a través de Internet, eliminando la necesidad de **infraestructura física local**.

## Principales Proveedores

### Amazon Web Services (AWS)
<img src="images/aws_icon.png" width="50px">

- **Ventaja**: Amplia gama de servicios y cobertura global.
- **Uso**: E-commerce, Big Data y streaming.

### Microsoft Azure

<img src="images/azure_icon.png" width="50px">

- **Ventaja**: Integración con productos Microsoft y entornos híbridos.
- **Uso**: Aplicaciones empresariales y DevOps (combina desarrollo *Dev* y operaciones *Ops* ).

### Google Cloud Platform (GCP)

<img src="images/google_cloud_icon.png" width="50px">

- **Ventaja**: Big Data, IA/ML y velocidad de infraestructura.
- **Uso**: Análisis de datos y Machine Learning.

## Modelos de Servicios

- **IaaS**: Infraestructura virtual (Ej: AWS EC2, GCE).
- **PaaS**: Plataformas de desarrollo (Ej: Azure App Services).
- **SaaS**: Software en la nube (Ej: Google Workspace).

## Máquinas Virtuales (VM) en la Nube

Las máquinas virtuales (VM) permiten ejecutar sistemas operativos y aplicaciones en un entorno virtualizado, optimizando el uso de servidores físicos mediante la virtualización y el uso de hipervisores.

### Proveedores Principales
- **GCP**: Google Compute Platform – Configuración flexible de recursos.
- **AWS**: Amazon EC2 – Instancias escalables y configurables.
- **Azure**: Azure Virtual Machines – Soporte Windows/Linux y autoescalado.

## Amazon SageMaker

Amazon SageMaker es un servicio de AWS que facilita la creación, entrenamiento y despliegue de modelos de Machine Learning en un entorno administrado.

### Tipos de Instancias SageMaker

- **Entrenamiento**: Optimizadas con GPUs para entrenar modelos complejos (Ej: ml.p3.2xlarge).
- **Inferencia**: Despliegue de modelos para predicciones en tiempo real o por lotes (Ej: ml.m5.large).
- **Notebook**: Entornos interactivos con Jupyter Notebooks para desarrollo y pruebas (Ej: ml.t2.medium).

### Casos de Uso
- Clasificación de imágenes.
- Procesamiento de lenguaje natural (NLP).
- Predicción de series temporales.

### Ventajas
- **Automatización**: AWS gestiona la infraestructura subyacente.
- **Escalabilidad**: Ajuste dinámico de recursos según la demanda.
- **Integración**: Compatible con servicios AWS como S3 y Lambda.

## Notebooks en la Nube

**Amazon SageMaker Notebooks**

<img src="images/SageMaker_icon.png" width="50px">
