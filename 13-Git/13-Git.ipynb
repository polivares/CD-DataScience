# Introducción a los sistemas Sistemas de Control de Versiones (VCS)

- Son sistemas que permiten llevar un registro sobre las **modificaciones** realizadas a un grupo de archivos pertenecientes a algún programa
- Permiten:
    - Revertir un archivo a un estado anterior.
    - Revertir un proyecto completo a su estado anterior.
    - Comparar cambios en el tiempo.
    - Ver quién realizó modificaciones sobre un archivo
    - etc.

# Tipos de VCS
## VCS Local 

- Aquellos que llevan un registro local de los archivos modificados utilizando una base de datos.
- Solo se permite acceder a las versiones de un archivo desde un computador (local)
- El más popular fue RCS (Revision Control System)
    - Guarda diferencias entre archivos de una versión a otra
    - Las guarda en un formato especial en disco
    - Recrea un archivo sumando al original las distintas diferencias almacenadas

<img src="images/VCSLocal.png" width="400px">

## VCS Centralizado

- Permiten realizar desarrollos colaborativos
- Existen un **único servidor** que contiene las versiones de cada archivo
- Se accede a través de un programa cliente
- Ventajas:
    - Permite determinar en que punto trabaja el resto de desarrolladores
    - Es más facil lidiar con una sola BD centralizada que con múltiples BD locales
- Desventajas:
    - Si el servidor falla, el sistema de versionamiento falla
    - Para trabajar, se necesita acceso al servidor
- El más popular es Subversion <img src="images/subversion_logo.svg" width="50px">


<img src="images/VCSCentralizado.png" width="400px">

## VCS Distribuido

- Cada cliente tiene una copia **integra** del repositorio
- Una descarga de una instatánea equivale a una copia completa de los datos
- Ventajas:
    - Si el servidor falla, existen múltiples copias remotas de seguridad
    -No es necesario tener acceso al servidor para trabajar
- Desventajas:
    - Se descargan todos los archivos del proyecto, lo que implica un mayor uso de espacio en disco duro
- Los más populares son GIT y Mercurial

# Componentes y Flujo de Trabajo en Git

## Los Tres Estados de los Archivos en Git
En GIT, los archivos pueden encontrarse en uno de los siguientes estados:

1. **Committed (Confirmado):**  
   - Los datos están almacenados de manera segura en la base de datos local.

2. **Modified (Modificado):**  
   - El archivo ha sido cambiado, pero aún no está preparado para ser almacenado en la base de datos.

3. **Staged (Preparado):**  
   - El archivo ha sido marcado para ser incluido en el próximo commit.

---

## Tres Secciones de un Proyecto Git

Un proyecto Git tiene tres áreas principales que gestionan los archivos:

1. **Directorio Git:**  
   - Contiene los metadatos y la base de datos del proyecto.  
   - Es donde Git almacena las instantáneas completas de los archivos.

2. **Directorio de Trabajo (Working Tree):**  
   - Es una copia de una versión del proyecto que se puede modificar.  
   - Los archivos en este directorio pueden ser editados o marcados para su inclusión.

3. **Área de Preparación (Staging Area):**  
   - Es un archivo sencillo que almacena información sobre los cambios que serán incluidos en el próximo commit.  

---

## Flujo de Trabajo en Git

El flujo de trabajo básico en Git sigue estos pasos:

1. **Modificar archivos en el directorio de trabajo:**  
   - Realiza cambios en los archivos necesarios.

2. **Preparar los archivos modificados (`Staging`):**  
   - Usa el comando `git add` para añadir los archivos modificados al área de preparación.

3. **Confirmar los cambios (`Commit`):**  
   - Usa el comando `git commit` para guardar una instantánea de los archivos en el directorio Git.

<img src="images/GITStates.png" width="500px">

### Ejemplo de Flujo de Trabajo Básico
```bash
# Crear un repositorio Git
git init

# Verificar el estado del repositorio
git status

# Modificar un archivo (puedes crear uno nuevo también)
echo "Este es un archivo de ejemplo" > archivo.txt

# Preparar el archivo para el commit
git add archivo.txt

# Confirmar los cambios
git commit -m "Se agregó el archivo de ejemplo"

# Verificar el historial de commits
git log
```

## Ciclo de Vida de un Archivo

Los archivos en GIT pueden estar en dos estados principales dentro del directorio de trabajo:

1. **Tracked (Bajo Seguimiento):**  
   - Archivos que ya están registrados en el sistema de control de versiones.  
   - Incluye tres sub-estados: `Committed`, `Modified`, `Staged`

2. **Untracked (Sin Seguimiento):**  
   - Archivos que no están incluidos en la última instantánea ni en el área de preparación.  
   - Pueden añadirse al control de versiones utilizando el comando `git add`.

## Ejercicio Práctico: Explorando los Estados y el Flujo de Trabajo

### Paso 1: Instalación de Git

1. **Instala Git en tu sistema operativo:**
   - En **Linux** (Debian/Ubuntu):
     ```bash
     sudo apt update
     sudo apt install git
     ```
   - En **macOS**:
     ```bash
     brew install git
     ```
   - En **Windows**:
     - Descarga e instala Git desde [git-scm.com](https://git-scm.com/).
     - Asegúrate de seleccionar las opciones predeterminadas en el instalador.

2. **Verifica que Git se instaló correctamente:**
  ```bash
      git --version
  ```

### Paso 2: Configuración Inicial

1. **Configura tu nombre de usuario y correo electrónico:**

```bash
git config --global user.name "Tu Nombre"
git config --global user.email "tuemail@example.com"
```

2. **Verifica la configuración:**
```bash
git config --list
```

### Paso 3: Crear y Configurar un Repositorio

1. **Crea una carpeta de trabajo y navega a ella:**

```bash
   mkdir mi_proyecto
   cd mi_proyecto
   ```

2. **Inicializa un repositorio GIT:**
   ```bash
   git init
   ```

3. **Verifica el estado del repositorio (debería estar vacío):**
   ```bash
   git status
   ```

---

### Paso 4: Flujo de Trabajo Básico

1. **Crea un archivo llamado `mi_archivo.txt` y añade contenido inicial:**
   ```bash
   echo "Este es el contenido inicial" > mi_archivo.txt
   ```

2. **Verifica el estado del archivo (`Untracked`):**
   ```bash
   git status
   ```

3. **Añade el archivo al área de preparación (`Staging`):**
   ```bash
   git add mi_archivo.txt
   ```

4. **Confirma los cambios con un mensaje:**
   ```bash
   git commit -m "Primer commit: Añadido mi_archivo.txt"
   ```

---

### Paso 5: Modificar y Confirmar Cambios

1. **Edita el archivo añadiendo una nueva línea:**
   ```bash
   echo "Nueva línea añadida al archivo" >> mi_archivo.txt
   ```

2. **Verifica el estado (`Modified`):**
   ```bash
   git status
   ```

3. **Revisa las diferencias entre la versión modificada y la última confirmada:**
   ```bash
   git diff
   ```

4. **Prepara los cambios para confirmar:**
   ```bash
   git add mi_archivo.txt
   ```

5. **Confirma los cambios:**
   ```bash
   git commit -m "Segundo commit: Modificada mi_archivo.txt"
   ```

---

### Paso 6: Explorar el Historial de Cambios

1. **Consulta el historial de commits:**
   ```bash
   git log
   ```

2. **Verifica los cambios en detalle:**
   ```bash
   git log -p
   ```

---

### Paso 7: Eliminar y Restaurar Archivos

1. **Elimina el archivo del directorio de trabajo:**
   ```bash
   rm mi_archivo.txt
   ```

2. **Verifica el estado del repositorio (`Deleted`):**
   ```bash
   git status
   ```

3. **Restaura el archivo eliminado desde el repositorio:**
   ```bash
   git checkout mi_archivo.txt
   ```

4. **Confirma que el archivo ha sido restaurado:**
   ```bash
   ls
   cat mi_archivo.txt
   ```

---

### Paso 8: Ignorar Archivos

1. **Crea un archivo `.gitignore` para excluir archivos innecesarios:**
   ```bash
   echo "*.log" > .gitignore
   echo "node_modules/" >> .gitignore
   ```

2. **Añade el archivo `.gitignore` al control de versiones:**
   ```bash
   git add .gitignore
   git commit -m "Añadido archivo .gitignore"
   ```

3. **Verifica que los archivos definidos en `.gitignore` son excluidos:**
   ```bash
   git status
   ```

# Introducción a GitHub 

<img src="images/github-mark.svg" width="50px"> <img src="images/GitHub_Logo.png" width="120px">

## ¿Qué es GitHub?

GitHub es una plataforma basada en la nube que permite a los desarrolladores almacenar, gestionar y colaborar en proyectos de software utilizando GIT. Ofrece herramientas para:

- **Control de versiones**: Gestión eficiente de cambios en el código.
- **Colaboración**: Trabajo en equipo mediante pull requests y revisiones.
- **Almacenamiento en la nube**: Repositorios públicos y privados.
- **Automatización y despliegue**: A través de GitHub Actions.

---

## Características Principales de GitHub

1. **Repositorios**:
   - Espacios para almacenar código y archivos relacionados.
   - Pueden ser públicos (accesibles para todos) o privados (solo accesibles con permisos).

2. **Forks**:
   - Copias de un repositorio que permiten a otros usuarios experimentar sin afectar el original.

3. **Pull Requests**:
   - Solicitudes para fusionar cambios de una rama o fork al repositorio principal.

4. **GitHub Actions**:
   - Herramientas de integración y despliegue continuo para automatizar flujos de trabajo.

5. **Wiki y Documentación**:
   - Espacio para añadir documentación sobre el proyecto.

6. **Issues y Proyectos**:
   - Seguimiento de errores, solicitudes de características y gestión del progreso.

# Actividad: Trabajo en Parejas con GitHub

## Objetivo

Aprender a colaborar en proyectos usando GitHub, desde la creación de cuentas hasta la resolución de conflictos al trabajar en el mismo archivo.

---

## Pasos de la Actividad

### 1. Crear Cuentas en GitHub

1. **Ambos participantes** deben crear una cuenta gratuita en [GitHub](https://github.com/).
2. Verifiquen sus correos electrónicos y completen la configuración inicial de sus cuentas.

---

### 2. Crear un Repositorio Compartido

1. Uno de los participantes (**Participante A**) crea un nuevo repositorio:
   - Ve a GitHub y haz clic en **New Repository**.
   - Asigna un nombre al repositorio, por ejemplo: `proyecto-colaborativo`.
   - Selecciona la opción de inicializar el repositorio con un archivo `README.md`.
   - Haz clic en **Create Repository**.

2. **Invitar al compañero**:
   - Ve a la pestaña **Settings** del repositorio.
   - Selecciona **Collaborators and teams** y añade el nombre de usuario de tu compañero (**Participante B**) como colaborador.

---

### 3. Clonar el Repositorio

1. Ambos participantes deben clonar el repositorio en sus computadoras:
   ```bash
   git clone <URL-del-repositorio>
    ```
    
### 4. Modificar el Repositorio y Subir Cambios

**Paso 1: Participante A añade un archivo inicial**
1. Crea un archivo llamado equipo.txt y añade los nombres de ambos participantes:

```bash
echo "Equipo Colaborativo: Participante A y Participante B" > equipo.txt
```
2. Añade y confirma los cambios:

```bash
git add equipo.txt
git commit -m "Añadido archivo equipo.txt con nombres"
git push origin main
```

**Paso 2: Participante B añade más información**

1. Descarga los últimos cambios del repositorio remoto:
```bash
git pull origin main
```

2. Edita el archivo equipo.txt añadiendo el propósito del equipo:
```bash
echo "Propósito: Aprender GitHub de manera colaborativa" >> equipo.txt
```
3. Añade y confirma los cambios:
```
git add equipo.txt
git commit -m "Añadido el propósito del equipo"
git push origin main
```

### 5. Resolver Conflictos

**Ambos participantes editan equipo.txt al mismo tiempo.**
1. Participante A edita equipo.txt añadiendo:
```bash
echo "Rol A: Gestor del Proyecto" >> equipo.txt
```
Y sube los cambios:
```bash
git add equipo.txt
git commit -m "Añadido Rol A"
git push origin main
```
2. Participante B edita equipo.txt añadiendo:
```bash
echo "Rol B: Desarrollador" >> equipo.txt
```

Al intentar subir los cambios, recibe un error:
```bash
git push origin main
```

**Resolución del Conflicto:**
1. Participante B descarga los cambios del repositorio remoto:
```bash
git pull origin main
```
2. Observa las marcas de conflicto en equipo.txt y edita el archivo para resolverlo. Ejemplo:
```bash
<<<<<<< HEAD
Rol B: Desarrollador
=======
Rol A: Gestor del Proyecto
>>>>>>> main
```
Resuelve como:
```bash
Rol A: Gestor del Proyecto
Rol B: Desarrollador
```
3. Añade y confirma la resolución:
```bash
git add equipo.txt
git commit -m "Resuelto conflicto en equipo.txt"
git push origin main
```


# Uso de Git con Visual Studio Code

## Configuración Inicial

Visual Studio Code (VS Code) incluye integración con Git, lo que permite gestionar repositorios de manera eficiente. Asegúrate de tener Git instalado en tu sistema y configurado previamente.

---

## Ver el Estado del Repositorio

1. Abre el proyecto en VS Code.
2. Haz clic en el icono de **Fuente de control** en la barra lateral izquierda (el ícono con una rama y un reloj).
3. VS Code mostrará los archivos modificados y no rastreados.

---

## Añadir Cambios al Área de Preparación

1. Selecciona los archivos que deseas añadir al área de preparación desde la lista en la sección **Cambios**.
2. Haz clic en el icono de `+` junto al archivo o en **+ Cambios** para añadir todos los archivos.

---

## Realizar un Commit

1. Después de añadir archivos al área de preparación, introduce un mensaje descriptivo en el cuadro de texto que aparece en la sección **Fuente de control**.
2. Haz clic en el ícono de check (`✔`) o presiona `Ctrl+Enter` (Windows/Linux) o `Cmd+Enter` (macOS) para confirmar el commit.

---

## Sincronizar Cambios con un Repositorio Remoto

1. Si el repositorio ya está vinculado a un remoto (por ejemplo, en GitHub), haz clic en el botón **Sincronizar Cambios** en la barra superior o en la sección **Fuente de control**.
2. Si no está vinculado, puedes configurarlo manualmente:
   - Abre el terminal integrado (Ctrl+`) y ejecuta:
     ```bash
     git remote add origin <URL-del-repositorio>
     git push -u origin main
     ```

---

## Realizar un Pull para Actualizar los Cambios

1. Haz clic en el botón **Sincronizar Cambios**, y VS Code realizará automáticamente un `git pull` antes de un `git push`.
2. Si ocurre un conflicto, VS Code resaltará las líneas afectadas en los archivos. Resuelve los conflictos directamente en el editor y realiza un nuevo commit.

---

## Vista de Historial

1. En la sección **Fuente de control**, haz clic derecho en un archivo y selecciona **Abrir comparación de cambios** para ver las diferencias entre versiones.
2. Puedes instalar extensiones como **Git Graph** para visualizar el historial de commits en forma de árbol interactivo.
